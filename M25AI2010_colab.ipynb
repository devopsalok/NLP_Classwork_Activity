## Smoothing to rescue!

```python
#Training a laplace smoothed unigram language model using training data

!./ngram-count -text /content/drive/MyDrive/SRILM/sample-data/sample-train.txt -order 1 -write /content/drive/MyDrive/SRILM/sample-data/files/sample-train-unigram-smoothed.count \
-lm /content/drive/MyDrive/SRILM/sample-data/files/sample-train-unigram-smoothed.lm -addsmooth 1 -unk


#Training a laplace smoothed bigram language model using training data

!./ngram-count -text /content/drive/MyDrive/SRILM/sample-data/sample-train.txt -order 2 -write /content/drive/MyDrive/SRILM/sample-data/files/sample-train-unigram-smoothed.count \
-lm /content/drive/MyDrive/SRILM/sample-data/files/sample-train-unigram-smoothed.lm -addsmooth 1 -unk
```

```python
#Testing the above unigram language model

!./ngram -lm /content/drive/MyDrive/SRILM/sample-data/files/sample-train-unigram-smoothed.lm -ppl /content/drive/MyDrive/SRILM/sample-data/test-q5.txt -debug 2

#Testing the above bigram language model

!./ngram -lm /content/drive/MyDrive/SRILM/sample-data/files/sample-train-unigram-smoothed.lm -ppl /content/drive/MyDrive/SRILM/sample-data/test-q5.txt -debug 2
```
